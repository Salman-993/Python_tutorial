```python
#Wap to find the greatest of 3 numbers entered by the user
a=int(input("Enter A"))
b=int(input("Enter b"))
c=int(input("Enter c"))

if(a>=b and a>c):
    print("A is the greatest among all",a)
elif(b>a and b>c):
    print("B is the greatest",b)
elif(c>a and c>b):
    print("C is the greatest number",c)
else:
    print("Invalid entery")
```
