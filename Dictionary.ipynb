```python
#Dictionaries are used to store data values in Key:Value pairs
#They are unordered,mutable and don't allow to duplicate keys
import numpy as np
dict = {
    "key" :"Value",
    "name" : "Salman",
    "learning" :"Python",
    "Age" : "21",
    "Subject" : ["List", "Python","C++"],
    "tup" : ("tuple"),
    12 : 43 ,
}
print(dict)
```

```python
print(dict["name"])
print(dict["Age"])
print(dict["learning"])
print(dict["Subject"])
```

```python
#We can also change values of key
#and can assign new key value
dict["Age"] = "34"
dict["surname"] = "Nazir"
print(dict)
```

```python
#We can also create null dictionary
null_dict = {}
print(null_dict)
# And can add keys and values in it timly
null_dict["name"] = "salman NAzir"
print(null_dict)
```

```python
# Nested Dictionaries
student = {
    "name": "Salman",
    "subjects" : {
        "Eng" : 91,
        "Math" : 78,
        "Phy" : 87,   
    }
}
print(student)
print(student["subjects"])
print(student["subjects"]["Phy"])
```
