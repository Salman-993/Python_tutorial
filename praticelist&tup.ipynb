```python
#Write a program to check if a list contains palindrome of elements.
# Palindrome means a list that is same if you reverse it 
# example [Ma'aM]  , [1,2,3,2,1] , [racecar]
list1 = [1,2,1,]
copy_list1 = list1.copy()
copy_list1.reverse()

if(copy_list1 == list1):
    print("Palindrome")
else:
    print("Not Palindrome")

```
