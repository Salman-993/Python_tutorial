```python
class Car:
    color = "Red"
    brand = "Toyota"
car1 = Car()
print(car1.color)  # Output: Red
print(car1.brand)  # Output: Toyota

car2 = Car()
print(car2.color)  # Output: Red
print(car2.brand)  # Output: Toyota
```

```python
class Student:
    def __init__(self, fullname ,marks):
        self.fullname = fullname
        self.marks = marks
        print("adding new student in database...")

s1 = Student("Adam", 85)
print(s1.fullname)
print(s1.marks)

s2 = Student("Zeus", 90)
print(s2.fullname , s2.marks)
```
