```python
    #Tuples
    # Tuple is a built  in data-type that lets us create immutable sequence of values
```

```python
tuple = (2,1,3,5,4,5,6,5)
print(type(tuple))
print(tuple[0])
print(tuple)
#if i want to create a single value tuple tup=(1,)i have to use , at it's end
tup=(1,)
print(tup)
#Slicing 
print(tuple[0:3])
#.index is used to find the index of specific value
print("The index of 5 is", tuple.index(5))
#tuple.count(el) Counts total number of occurance 
print(tuple.count(5))
```
