```python
#Types of while loop
#Break : used to terminate the loop when encountered.
#Continue: Terminate execution in the current iteration and continue execution of the loop with the next iteration.
```

```python
#Break

i=0
while i <=5:
    print(i)
    if(i ==3):
        break
    i += 1
print("End of Loop")
```

```python
#Break
tu = (1,4,9,16,25,36,49,64,81,100)
i = 0
x = 100
while(i <=len(tu)):
    if(tu[i] == x):
        print("Found",x , "at", i , "Index")
        break
    else:
        print("Finding...")
    i +=1

```

```python
#Continue
i = 0
while i<=5:
    if(i == 4):
        i += 1
        continue #Skip thing after it
    print(i)
    i += 1
```

```python
#Continue for odd number
i = 0
while i <= 10:
    if(i % 2 == 0):
        i += 1
        continue #Skip thing after it
    print(i)
    i += 1
```

```python
#Continue for even number
i = 0
while i <= 10:
    if(i % 2 != 0):
        i += 1
        continue #Skip thing after it
    print(i)
    i += 1
```
