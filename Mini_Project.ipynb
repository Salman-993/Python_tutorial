Guess The Number

```python
import random

randNum = random.randint(1,100)


while True:
    guess = int (input("Guess a number between 1 and 100:"))
    if guess == randNum:
        print("Congratulations! You guessed the number.", randNum)
        break
    elif guess < randNum:
        print("The number you entered is to low. Try again.")
    else:
        print("The number you entered is to high. Try again.")

print("Game Over.")
```
