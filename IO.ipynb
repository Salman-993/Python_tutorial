```python
#With Syntax
with open("demo.txt" , "r") as f:
    data = f.read()
    print(data)

with open("demo.txt" , "w") as f:
    f.write("This is the new line using with syntax")
    print(data)
```

```python
#Write a program to replace the occurrences of Java with python in above file
with open("practice.txt","r") as f:
    data = f.read()

new_data=data.replace("java","python")
print(new_data)

with open("practice.txt","w") as f:
    f.write(new_data)
```

```python
#Search if the word "learning " exist in the file or not
word = "learning"
with open("practice.txt","r") as f:
    data = f.read()
    if (data.find(word) != -1):
        print("word found")
    else:
        print("word not found")
```
