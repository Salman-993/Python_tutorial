```python
#Slicing
#Accessing parts of a string
#str[start:stop:step]
str ="Lahore Pakistan"
print(str[0:7])
print(str[7:len(str)])
print(str[0:7])#[0:7]
print(str[2:])#[2:end]
#Minus Indexing
a="Apple"
print(a[-5:-2])
print(a[-5:5])
```

```python
marks=int(input("Enter Student marks:"))
if(marks>=90 and marks<=100):
    grade = "A"
    print("Student grade is ",grade)
elif(marks>=80 and marks<90):
    grade = "B"
    print("Student grade is ",grade)
elif(marks>=65 and marks<80):
    grade = "C"
    print("Student grade is ",grade)
elif(marks>=50 and marks<65):
    grade = "D"
    print("Student grade is ",grade)
elif(marks<50):
    grade = "F"
    print("Student grade is ",grade)
else:
    print("Invalid marks")


```
